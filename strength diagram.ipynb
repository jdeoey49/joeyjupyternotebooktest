# Gradient Method ($S=\frac{x+y}{2}$)

```python
import numpy as np
import matplotlib.pyplot as plt

# Example candidate points
technical_score = np.array([64, 66, 71, 71, 51])
economical_score = np.array([65, 65, 70, 63, 50])
labels = ['Joey', 'Gulce', 'Jonathan', 'Hasan', 'Pinar']

# Different marker and color for each point
markers = ['D', '^', 'o', 's', '*']
colors = ['red', 'blue', 'limegreen', 'orange', 'purple']

# Manual label offsets for each point
offsets = [
    (6, 8),      # Joey
    (-10, -14),    # Gulce
    (8, 10),     # Jonathan
    (8, -16),    # Hasan
    (6, 6)       # Pinar
]

# Create grid for contour plot
x = np.linspace(0, 100, 200)
y = np.linspace(0, 100, 200)
X, Y = np.meshgrid(x, y)

# Strength function
S = (X + Y) / 2

# Plot
plt.figure(figsize=(8, 6), facecolor='white')
ax = plt.gca()
ax.set_facecolor('white')

# Contour lines
lines = plt.contour(X, Y, S, levels=10, colors='black', linewidths=0.7, alpha=0.6)
plt.clabel(lines, inline=True, fontsize=8, colors='black')

# Scatter points and labels
for x_val, y_val, label, marker, color, offset in zip(
    technical_score, economical_score, labels, markers, colors, offsets
):
    plt.scatter(
        x_val, y_val,
        marker=marker,
        c=color,
        edgecolors='black',
        s=120,
        zorder=3
    )

    plt.annotate(
        label,
        (x_val, y_val),
        textcoords="offset points",
        xytext=offset,
        fontsize=9,
        bbox=dict(facecolor='white', edgecolor='none', alpha=0.8, pad=1.5)
    )

# Labels and title
plt.xlabel('Technical Score')
plt.ylabel('Economical Score')
plt.title(r'Gradient Methods: $s=\frac{x+y}{2}$')
plt.xlim(0, 100)
plt.ylim(0, 100)
plt.grid(alpha=0.2)

plt.show()
```

# Hyperbola Method ($s=\sqrt{xy}$)

```python
import numpy as np
import matplotlib.pyplot as plt

technical_score = np.array([64, 66, 71, 71, 51])
economical_score = np.array([65, 65, 70, 63, 50])
labels = ['Joey', 'Gulce', 'Jonathan', 'Hasan', 'Pinar']
colors = ['red', 'blue', 'limegreen', 'orange', 'purple']
markers = ['D', '^', 'o', 's', '*']

# Manual label offsets in points: (x_offset, y_offset)
offsets = {
    'Joey': (-18, 20),
    'Gulce': (-20, -14),
    'Jonathan': (10, 10),
    'Hasan': (10, -16),
    'Pinar': (10, 8)
}

# Grid from 0 to 100
x = np.linspace(1, 100, 400)
y = np.linspace(1, 100, 400)
X, Y = np.meshgrid(x, y)

# Hyperbola-style strength function
S = np.sqrt(X * Y)

plt.figure(figsize=(7, 7))

levels = np.arange(10, 101, 10)
contours = plt.contour(X, Y, S, levels=levels, colors='black', linewidths=1.2)
plt.clabel(contours, inline=True, fontsize=9, fmt='s = %.0f')

for tx, ey, label, color, marker in zip(
    technical_score, economical_score, labels, colors, markers
):
    plt.scatter(
        tx, ey,
        s=300,
        color=color,
        marker=marker,
        edgecolors='black',
        zorder=3
    )

    dx, dy = offsets[label]
    plt.annotate(
        label,
        xy=(tx, ey),
        xytext=(dx, dy),
        textcoords='offset points',
        fontsize=11,
        ha='left',
        va='center',
        bbox=dict(facecolor='white', edgecolor='none', alpha=0.8, pad=1.5)
    )

plt.xlim(0, 100)
plt.ylim(0, 100)
plt.xlabel('Technical Score x')
plt.ylabel('Economical Score y')
plt.title(r'Hyperbola Method: $s=\sqrt{x \cdot y}$')
plt.grid(False)
plt.gca().set_aspect('equal', adjustable='box')

plt.tight_layout()
plt.show()
```
