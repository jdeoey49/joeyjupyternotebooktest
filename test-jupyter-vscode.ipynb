```python
interest = float(input("Enter the interest rate: "))
initial_investment = int(input("Enter the initial investment: "))
years = int(input("Enter the number of years: "))
liquiditation_value = int(input("Enter the liquidation value: "))
NPV = 0;
sum_dcf = 0;
for i in range(1, years+1):
    cf = int(input("Enter the cash flow for year " + str(i) + ": "))
    dcf = cf / (1 + interest)**i
    sum_dcf += dcf
discounted_liquidation_value = liquiditation_value / (1 + interest)**years
NPV = sum_dcf + discounted_liquidation_value - initial_investment
print("The NPV of the project is: " ,NPV)
```
