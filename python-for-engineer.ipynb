# Using Python as a Calculator

```python
height = 15.5
width = 10.2
area = height * width
print("The area of the rectangle is", area)
```

```python
cath1 = 4
cath2 = 3
hypotenuse = (cath1**2 + cath2**2)**0.5 # ** is the exponentiation operator
print("The length of the hypotenuse is", hypotenuse)
```

```python
vat = 19/100
netPrice = 255.6
netPrice * vat
```

```python
grossPrice = netPrice + _ # _ is the result of the last expression, which is netPrice * vat
# _ stores value of last expression, not explicitly assigned to a variable
# in jupyter notebook, code is executed cell by cell, so netPrice * vat is executed before grossPrice = netPrice + _ and the result of netPrice * vat is stored in _ and can be used in the next cell.
print("The gross price is", grossPrice)
```

```python
print("The gross price is", round(_, 2))
```
